```python
def findDog(s):
    return 'dog' in s
```

```python
print(findDog("Is there a dog here?"))
```

```python
def countDog(s):
    return s.split().count('dog')
```

```python
print(countDog("This dog runs faster than the other dog"))
```

```python
def caught_speeding(speed,is_birthday):
    if is_birthday:
        speed -= 5
    if speed <= 60:
        return "No Ticket"
    elif speed <= 80:
        return "Small Ticket"
    else:
        return "Big Ticket"
```

```python
print(caught_speeding(60,False))
```

```python
print(caught_speeding(65,True))
```

```python
print(caught_speeding(80,False))
```

```python
print(caught_speeding(85,True))
```

```python
print(caught_speeding(90,False))
```

```python
print(caught_speeding(90,True))
```
