## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: your code here
# (a) Missing values: count per column, drop no Brand/Model, fill text columns with "Unknown"
print("T3a missing before:")
print(df.isna().sum())
rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
rows_removed = rows_before - len(df)
df["Registered City"] = df["Registered City"].fillna("Unknown")
df["Fuel"] = df["Fuel"].fillna("Unknown")
df["Transaction Type"] = df["Transaction Type"].fillna("Unknown")
print("T3a rows removed:", rows_removed)

# (b) Formatting: strip/title city names, remove "Rs" and commas from Price
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = df["Price"].str.replace("Rs", "").str.replace(",", "").str.strip()
print("T3b sample Price after cleaning text:", df["Price"].head(3).tolist())

# (c) Data types: Price -> int64, Year -> Int64
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c dtypes:")
print(df.dtypes)

# (d) Inconsistent categories: Sukkar -> Sukkur, Corrolla Altis -> Corolla Altis
cities_before = df["Registered City"].nunique()
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
cities_after = df["Registered City"].nunique()
print("T3d cities before:", cities_before, "| after:", cities_after)

# (e) Duplicates: count, drop, print new row count
print("T3e duplicate ads:", df.duplicated().sum())
df = df.drop_duplicates()
print("T3e new row count:", len(df))

# (f) Invalid values: impossible KMs -> NaN; drop fake prices and Suzuki/Daihatsu > 5,000,000
bad_kms = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | (df["KMs Driven"].isin([12345, 123456]))
df.loc[bad_kms, "KMs Driven"] = np.nan

bad_price = df["Price"].isin([12_345_678, 87_654_321]) | (df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000))
df = df[~bad_price]
print("T3f bad KMs set to NaN:", bad_kms.sum(), "| invalid price rows dropped:", bad_price.sum())

# (g) Self-check: final shape must be (20484, 9)
print("T3g final shape:", df.shape)
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: your code here
# (a) New column Price_lakh
df["Price_lakh"] = df["Price"] / 100_000
print("T4a sample Price_lakh:", df["Price_lakh"].head(3).tolist())

# (b) New column Age = 2020 - Year
df["Age"] = 2020 - df["Year"]
print("T4b sample Age:", df["Age"].head(3).tolist())

# (c) The 5 most common brands
print("T4c:")
print(df["Brand"].value_counts().head(5))

# (d) The 5 cities with the most ads
print("T4d:")
print(df["Registered City"].value_counts().head(5))

# (e) Median Price_lakh for each fuel type, highest first
print("T4e:")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# (f) Median Price_lakh of a Corolla GLI for each year 2010-2018
gli = df[(df["Model"] == "Corolla GLI") & (df["Year"] >= 2010) & (df["Year"] <= 2018)]
gli_median = gli.groupby("Year")["Price_lakh"].median()
print("T4f:")
print(gli_median)
avg_loss = (gli_median.loc[2018] - gli_median.loc[2010]) / 8
print("T4f average value lost per year (lakh):", round(avg_loss, 2))

# (g) The 5 most expensive ads
print("T4g:")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "Price", "Price_lakh", "Registered City", "Condition"]])

# (h) Number of ads and median price for Cash vs Installment/Leasing
print("T4h count:")
print(df.groupby("Transaction Type")["Price_lakh"].count())
print("T4h median (lakh):")
print(df.groupby("Transaction Type")["Price_lakh"].median())
```

**T4 Written Answers:**
- **(f) Corolla GLI depreciation (2010–2018):** The median price drops from **23.20 lakh** (2018) to **12.25 lakh** (2010), a total loss of **10.95 lakh** over 8 years — about **1.37 lakh per year** on average (around Rs 137,000/year), with steeper drops in newer years (~2.0–2.2 lakh/year from 2016–2018).
- **(g) Five most expensive ads:** Yes, the **2015 Toyota Prius** listed at **Rs 35,750,000 (357.5 lakh)** clearly looks wrong because a 2015 Prius normally sells for ~25 lakh (likely an extra trailing zero typed on Rs 3,575,000 / 35.75 lakh). The 2005 Land Cruiser (520 lakh) and 2007 Surf (410 lakh) also look inflated by an extra trailing zero compared to typical ~50–60 lakh prices, whereas the 2017 Land Cruiser (670 lakh) is a plausible high-end SUV price.
- **(h) Cash vs Installment/Leasing:** There are **19,634 Cash** ads (median **6.9 lakh**) versus **850 Installment/Leasing** ads (median **3.0 lakh**). Installment/Leasing ads have a much lower median price because sellers typically enter the down payment or advance amount in the `Price` field rather than the full vehicle price.
